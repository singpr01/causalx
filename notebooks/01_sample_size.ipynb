```python
from causalx.sampling import sample_size
```

```python
res_bin = sample_size(
    metric_type="binary",
    baseline_rate=0.10,
    mde=0.02,  # absolute lift: 10% -> 12%
    alpha=0.05,
    power=0.80,
    two_sided=True,
    allocation=0.50,
)

res_bin
```

```python
res_cont = sample_size(
    metric_type="continuous",
    sigma=1.5,  # assumed SD of the metric
    mean_diff=0.20,  # difference in means to detect
    alpha=0.05,
    power=0.80,
    two_sided=True,
    allocation=0.50,
)

res_cont
```

```python
res_70_30 = sample_size(
    metric_type="continuous",
    sigma=1.5,
    mean_diff=0.20,
    allocation=0.70,  # 70% treatment / 30% control
)

res_70_30
```

```python
def summarize(res):
    return {
        "metric_type": res.metric_type,
        "n_control": res.n_control,
        "n_treatment": res.n_treatment,
        "n_total": res.n_total,
        "alpha": res.alpha,
        "power": res.power,
        "two_sided": res.two_sided,
        "allocation": res.allocation,
        "assumptions": res.assumptions,
        "method": res.method,
    }


summarize(res_bin), summarize(res_cont), summarize(res_70_30)
```
